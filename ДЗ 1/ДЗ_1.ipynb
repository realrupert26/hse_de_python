## Задание 1

Переведите содержимое файла purchase_log.txt в словарь purchases вида: {'1840e0b9d4': 'Продукты', ...}

```python
import json

purchases= {}

with open('purchase_log.txt', mode='r', encoding='UTF-8') as f:
    next(f)
    for line in f:
        data = json.loads(line)
        purchases[data['user_id']] = data['category']
```

```python
for key, value in list(purchases.items())[:2]:
    print(key, "'" + str(value) + "'")
```

## Задание 2

Для каждого user_id в файле visit_log.csv определите третий столбец с категорией покупки. Если покупка была, сам файл visit_log.csv изменять не надо.

Запишите в файл funnel.csv визиты из файла visit_log.csv, в которых были покупки с указанием категории.

```python
import csv

with open('visit_log.csv', mode='r', encoding='UTF-8') as initial_data,\
     open('funnel.csv', mode='w', encoding='UTF-8') as final_data:
        visit_log = csv.reader(initial_data)
        funnel = csv.writer(final_data)
        funnel_head = ['user_id', 'source', 'category']
        funnel.writerow(funnel_head)
        for user_id, source in visit_log:
            category = purchases.get(user_id)
            if category is not None:
                funnel.writerow([user_id, source, category])
```
